```python
import numpy as np
import astropy.units as u

from simulate_rate_map import SimulateRateMap
```

# Pyechelle

```python
srm = SimulateRateMap()
```

```python
fiber_1_wavelengths = np.arange(0.9,2.46,0.01) * u.micron
fiber_1_flux = np.ones_like(fiber_1_wavelengths.value)*1 * u.erg/u.s/u.cm**2/u.micron
```

```python
srm.set_fiber_spectrum(fiber_number=1,
                       spectrum_wavelengths=fiber_1_wavelengths,
                       spectrum_flux=fiber_1_flux,
                      continuous=False)
```

```python
srm.output_directory = 'output/'
```

```python
srm.create_pyechelle_rate_map(spectrograph='BSPEC')
```

# Native simulator
